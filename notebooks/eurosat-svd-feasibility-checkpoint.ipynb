```python

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn

print("NumPy:", np.__version__)
print("scikit-learn:", sklearn.__version__)

print("\nAttached Kaggle datasets:")
for path in Path("/kaggle/input").iterdir():
    print(path)

print("\nSearching for EuroSAT class folders...")
for path in Path("/kaggle/input").rglob("AnnualCrop"):
    if path.is_dir():
        print("Found:", path.parent)
        print("Available folders:",
              sorted(p.name for p in path.parent.iterdir()
                     if p.is_dir()))
```

```python

from pathlib import Path
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# EuroSAT RGB dataset path
DATASET_PATH = Path(
    "/kaggle/input/datasets/apollo2506/eurosat-dataset/EuroSAT"
)

# Four classes from our proposal
CLASSES = ["AnnualCrop", "Forest", "Residential", "River"]

# Initial feasibility sample
IMAGES_PER_CLASS = 200
RANDOM_SEED = 42

rng = np.random.default_rng(RANDOM_SEED)

images = []
labels = []

for class_name in CLASSES:
    class_path = DATASET_PATH / class_name

    # Find images in class folder
    image_files = sorted(
        list(class_path.glob("*.jpg")) +
        list(class_path.glob("*.png")) +
        list(class_path.glob("*.jpeg"))
    )

    assert len(image_files) >= IMAGES_PER_CLASS, (
        f"Not enough images in {class_name}"
    )

    # Randomly select 200 images
    selected_files = rng.choice(
        image_files,
        size=IMAGES_PER_CLASS,
        replace=False
    )

    for file_path in selected_files:
        with Image.open(file_path) as img:
            img = img.convert("RGB").resize((64, 64))
            images.append(np.asarray(img, dtype=np.uint8))
            labels.append(class_name)

images = np.stack(images)
labels = np.array(labels)

print("Images shape:", images.shape)
print("Labels shape:", labels.shape)
print("Classes:", np.unique(labels, return_counts=True))

# Display one sample from each class
fig, axes = plt.subplots(1, 4, figsize=(12, 3))

for i, class_name in enumerate(CLASSES):
    idx = np.where(labels == class_name)[0][0]
    axes[i].imshow(images[idx])
    axes[i].set_title(class_name)
    axes[i].axis("off")

plt.tight_layout()
plt.show()
```

```python

from sklearn.model_selection import train_test_split
import numpy as np

# 1. Normalize RGB images to [0, 1]
X = images.astype(np.float32) / 255.0

# 2. Flatten each image into 12,288 features
X = X.reshape(len(X), -1)

# 3. Stratified train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    labels,
    test_size=0.20,
    random_state=42,
    stratify=labels
)

# 4. Verify the data
print("Full dataset:", X.shape)
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nPixel range:", X.min(), "to", X.max())

print("\nTraining class distribution:")
for class_name in CLASSES:
    print(class_name, ":", np.sum(y_train == class_name))

print("\nTesting class distribution:")
for class_name in CLASSES:
    print(class_name, ":", np.sum(y_test == class_name))

print("\nTraining memory:",
      round(X_train.nbytes / (1024**2), 2), "MB")

print("Testing memory:",
      round(X_test.nbytes / (1024**2), 2), "MB")
```

```python

import time
import numpy as np
from sklearn.decomposition import PCA
from sklearn.metrics import mean_squared_error

# Baseline rank from our proposal
K = 20

print("Starting PCA/SVD baseline...")
start_time = time.perf_counter()

# Fit PCA using randomized SVD
pca = PCA(
    n_components=K,
    svd_solver="randomized",
    random_state=42
)

pca.fit(X_train)

# Project test images into lower dimensions
X_test_reduced = pca.transform(X_test)

# Reconstruct test images
X_test_reconstructed = pca.inverse_transform(
    X_test_reduced
)

elapsed_time = time.perf_counter() - start_time

# Calculate evaluation metrics
mse = mean_squared_error(
    X_test,
    X_test_reconstructed
)

# Relative reconstruction error for each test image
numerator = np.linalg.norm(
    X_test - X_test_reconstructed,
    axis=1
)

denominator = np.linalg.norm(
    X_test,
    axis=1
)

relative_error = np.mean(
    numerator / np.maximum(denominator, 1e-12)
)

# Training explained variance
explained_variance = np.sum(
    pca.explained_variance_ratio_
)

print("\n===== BASELINE RESULTS =====")
print(f"Rank (k): {K}")
print(f"Training images: {len(X_train)}")
print(f"Testing images: {len(X_test)}")
print(f"Reconstruction MSE: {mse:.6f}")
print(f"Mean relative error: {relative_error:.6f}")
print(f"Cumulative explained variance: "
      f"{explained_variance * 100:.2f}%")
print(f"Execution time: {elapsed_time:.2f} seconds")

print("\nReduced test shape:", X_test_reduced.shape)
print("Reconstructed test shape:",
      X_test_reconstructed.shape)
```

```python

import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(
    2, 4,
    figsize=(13, 6)
)

for i, class_name in enumerate(CLASSES):
    # First test image belonging to this class
    idx = np.where(y_test == class_name)[0][0]

    original = X_test[idx].reshape(64, 64, 3)

    reconstructed = (
        X_test_reconstructed[idx]
        .reshape(64, 64, 3)
    )

    # Clip reconstructed values for image display only
    reconstructed_display = np.clip(
        reconstructed, 0, 1
    )

    axes[0, i].imshow(original)
    axes[0, i].set_title(f"{class_name}\nOriginal")
    axes[0, i].axis("off")

    axes[1, i].imshow(reconstructed_display)
    axes[1, i].set_title(f"{class_name}\nReconstructed (k=20)")
    axes[1, i].axis("off")

plt.suptitle(
    "EuroSAT RGB — PCA/SVD Baseline Reconstruction",
    fontsize=14
)

plt.tight_layout()
plt.savefig(
    "/kaggle/working/baseline_reconstruction.png",
    dpi=200,
    bbox_inches="tight"
)
plt.show()
```

```python

import pandas as pd

baseline_results = pd.DataFrame([{
    "Dataset": "EuroSAT RGB",
    "Training Images": len(X_train),
    "Testing Images": len(X_test),
    "Rank": K,
    "MSE": mse,
    "Mean Relative Error": relative_error,
    "Explained Variance": explained_variance,
    "Runtime Seconds": elapsed_time,
    "Random Seed": 42
}])

baseline_results.to_csv(
    "/kaggle/working/baseline_results.csv",
    index=False
)

display(baseline_results)

print("Saved baseline_results.csv")
print("Saved baseline_reconstruction.png")
```
